```python
# Download stock data (daily) for your firm
df = yf.download("PEG", period = "5y", interval = "1d")
```

```python
# Calculate Returns
rtrn = df["Adj Close"].pct_change()
```

```python
# Calculate return autocorrelation
rtrn.autocorr()
```

```python
# Plot a histogram of returns
sns.displot(rtrn)
```
